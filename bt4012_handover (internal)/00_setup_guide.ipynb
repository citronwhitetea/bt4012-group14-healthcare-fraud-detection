# Setup Guide — BT4012 Healthcare Provider Fraud Detection

Read this before touching any other notebook. It covers: cloning the repo, setting up your Python environment, understanding the folder structure, getting the data, and how we work together in Git.

You only need to do this setup **once** per machine.

## 1. Clone the repo

In your terminal (Terminal on Mac, or VS Code's integrated terminal):

```bash
git clone https://github.com/citronwhitetea/bt4012-group14-healthcare-fraud-detection.git
cd bt4012-group14-healthcare-fraud-detection
code .
```

`code .` opens the folder in VS Code with the terminal already pointed at the repo. If `code` isn't recognized, open VS Code manually, run Cmd+Shift+P -> "Shell Command: Install 'code' command in PATH", then try again.

## 2. Create your own virtual environment

Each person needs their **own** local environment — it is not shared or committed to the repo (see `.gitignore`).

```bash
python3 -m venv .venv
source .venv/bin/activate        # Windows: .venv\Scripts\activate
pip install -r requirements.txt
```

Your terminal prompt should now show `(.venv)` at the front. In VS Code, run Cmd+Shift+P -> "Python: Select Interpreter" -> choose the one inside `./.venv/`.

If you install a new package while working, remember to update `requirements.txt` for everyone else:
```bash
pip freeze > requirements.txt
```

## 4. Folder structure — what goes where

```
data/raw/            <- put the 4 downloaded CSVs here
data/processed/      <- cleaned/merged data your notebooks generate (auto-created by scripts)
notebooks/            <- one notebook per pipeline stage; this is where you run and explore
src/                  <- shared functions imported into notebooks (data loading, cleaning, features, modeling)
reports/figures/      <- exported plots for the final report
```

**Golden rule:** if you write a function you expect to reuse (e.g. a cleaning step, a feature aggregation), put it in `src/`, then import it into your notebook:

```python
from src.data_loader import load_all_tables
from src.cleaning import clean_beneficiary_data
```

This keeps everyone using the same logic instead of quietly diverging across notebooks.

## 6. Git workflow

- Pull before you start working: `git pull`
- Prefer working on a branch per feature/section rather than directly on `main`:
  ```bash
  git checkout -b yourname-eda
  ```
- Commit often with clear messages:
  ```bash
  git add .
  git commit -m "Add provider-level aggregation for claim counts"
  git push -u origin yourname-eda
  ```
- Open a Pull Request on GitHub to merge into `main` once your section works, so the group can review before merging.
- If you're only editing your own notebook and nobody else touches it, working directly on `main` with frequent small commits is also fine — just pull first to avoid conflicts.

**Never commit:** your `.venv/` folder, `.env` files, or large generated artifacts — these are already excluded in `.gitignore`.
